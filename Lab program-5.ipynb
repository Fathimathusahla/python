```python
n=int(input("Enter a number:"))
for i in range(1,n+1):
    for j in range(1,i+1):
        print(i*j,end=" ")
    print()
```

```python
square= lambda a:a*a
triangle=lambda l,b:0.5*l*b
rectangle=lambda b,h:b*h
while True:
    print("\n1.square")
    print("2.Triangle")
    print("3.Rectangle")
    print("4.exit")
    ch=int(input("Enter your choice:"))
    if (ch==1):
        a=int(input("Enter value:"))
        print("Area",square(a))
    elif(ch==2):
        l=int(input("Enter the length:"))
        b=int(input("Enter the height:"))
        print("Area of trinagle",triangle(l,b))
    elif (ch==3):
        b=int(input("Enter the base:"))
        h=int(input("Enter the height:"))
        print("Area of trinagle",rectangle(b,h))
    elif(ch==4):
        print("Exit")
        break
    else:
        print("Invalid choice")
```
