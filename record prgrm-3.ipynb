```python
names=['aju','anu','akku','ansifa']
count=0
for name in names:
    count=count+name.count('a')
print("count of a:",count)
```

```python
color_list1=['RED','BLUE','GREEN']
color_list2=['BLUE','YELLOW']
for color in color_list1:
    if color not in color_list2:
        print(color)
```
